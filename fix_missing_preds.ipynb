```python
import pandas as pd
import os
```

```python
l = []
for path in os.listdir('data'):
    if path.endswith('.wav'):
        l.append({
            "key": path.removesuffix('.wav'),
            "source": os.path.join('/workspace/VideoLLM/data/', path),
            "target": "Not Available",}
)
```

```python
df = pd.DataFrame(l)
```
